# 04 Publication Tables And Figures

This notebook consolidates the high-signal upstream outputs into a small paper-facing table set and inventories the expected figures.

**Inputs.** It reads selected tables from Notebooks 0-3 and checks for expected figure files from Notebooks 1-3.

**Outputs.** It writes final manuscript table CSV/Markdown/LaTeX exports and three inventory CSVs under `outputs/*/04_publication_tables_figures/`.

**Key decisions.** Notebook 4 does not regenerate analytical figures. Figures remain owned by the notebooks that create the underlying analysis, while Notebook 4 records whether those expected files exist.

**When to rerun.** Rerun after any upstream notebook changes its final tables or figures, especially after switching Notebook 2 or Notebook 3 from smoke mode to full model mode.


## 1. Imports And Paths

Load config and identify the output folders that will be checked. The helper uses the same expected artifact registry as the tests, so missing final figures or upstream tables become visible in one place.


```python
from pathlib import Path
import sys

# Keep notebook imports stable whether the notebook is run from JupyterLab,
# VS Code, or the repository root.
article_root = Path.cwd()
while article_root.name != "2_journal_article":
    if article_root.parent == article_root:
        raise RuntimeError("Could not locate publication/2_journal_article")
    article_root = article_root.parent
notebook_dir = article_root / "notebooks"
if str(notebook_dir) not in sys.path:
    sys.path.insert(0, str(notebook_dir))

import _experiment_helpers as h

cfg = h.load_config(article_root)
paths = h.article_paths(article_root, cfg)
h.ensure_output_dirs(paths)
print(f"Article root: {article_root}")
print(f"Config schema: {cfg['schema_version']}")
print(f"Output root: {paths.outputs}")

print(paths.tables)
print(paths.figures)
```
