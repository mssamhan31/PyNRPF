# Minimal Bridge Method Ladder

This notebook tests thirteen low-parameter bridge-style day-level RPF detectors. The first three reuse existing daily bridge-score artifacts; E4-E6 use v0.3-style daily scores; E7-E8 add simple site-relative transforms of E4; E9-E10 compute E4 after input-level site scaling; E11-E13 add sustained-window duration guards/penalties. It is designed for fast iteration: no model training, no figures, and CSV outputs only.

## Experiments

1. `E1_bridge_only`: `bridge_ratio_p99` with one Alpha-selected global threshold.
2. `E2_bridge_plus_roughness`: `bridge_ratio_p99 + full_tv_ratio_p99` with one Alpha-selected global threshold.
3. `E3_bridge_plus_site_median`: `bridge_ratio_p99 - 0.425 * within_site_median(bridge_ratio_p99)` with one Alpha-selected global threshold.
4. `E4_v03_three_feature`: fast candidate-level max of `bridge + roughness + slope_continuity`.
5. `E5_v03_without_slope`: fast candidate-level max of `bridge + roughness`.
6. `E6_v03_three_feature_site_median`: E4 with one fixed site-median normalisation weight.
7. `E7_v03_three_feature_site_centered`: E4 minus each site's median E4 score.
8. `E8_v03_three_feature_site_rank`: each day's within-site E4 percentile rank.
9. `E9_v03_site_solar_scaled`: E4 computed after scaling net load and solar by each site's median daytime solar p95.
10. `E10_v03_site_combined_scaled`: E4 computed after scaling net load and solar by each site's median combined daytime solar/net-load scale.
11. `E11_v03_duration_ge_1h`: E4 with a selected-window duration guard of at least 1 hour.
12. `E12_v03_duration_ge_1p5h`: E4 with a selected-window duration guard of at least 1.5 hours.
13. `E13_v03_soft_duration_penalty`: E4 with a fixed soft penalty for selected windows shorter than 2.5 hours.

## Outputs

CSV files are written to `outputs/11_minimal_bridge_method_ladder/` beside this notebook. The evaluation reports pooled, macro-site-average, and per-site day-level precision, recall, and F1 for Alpha, Beta, and Beta sure-only days. When possible, the script reuses cached E4/E5 daily scores, selected-window metadata, and scaled E9/E10 scores from the previous joined-score CSV.
