# E4 Beta FP/FN Visual Examples

This notebook creates a fast visual diagnostic for `E4_v03_three_feature`, focusing only on Beta day-level false positives and false negatives from the latest minimal bridge ladder outputs.

It does not train models and does not rescan the full dataset. By default this wrapper runs the sure-only, per-site/per-confusion-group batch: one HTML per Beta site and FP/FN group, sorted by date. The script can still be run in the old top-example mode by changing the environment variables.

Each panel shows raw net load, solar generation, the E4 selected sign-flipped window, the manual reviewed window, and the E4 component scores: bridge, roughness, slope-continuity, total score, and Alpha-selected threshold.

```python
import os
import runpy
from pathlib import Path

candidates = [
    Path.cwd() / "12_e4_beta_fp_fn_visual_examples.py",
    Path.cwd() / "publication/2_journal_article/notebooks/99_Misc/12_e4_beta_fp_fn_visual_examples.py",
]
os.environ["E4_EXAMPLE_MODE"] = "site_group_all"
os.environ["E4_EXAMPLE_SUBSET"] = "sure_only"

script = next((path for path in candidates if path.exists()), None)
if script is None:
    raise FileNotFoundError("Could not find 12_e4_beta_fp_fn_visual_examples.py from the current working directory")

runpy.run_path(str(script), run_name="__main__")
```
